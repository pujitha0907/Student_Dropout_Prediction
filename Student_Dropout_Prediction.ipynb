```python
# Import pandas for dataset handling
import pandas as pd

# Import numpy for numerical operations
import numpy as np

# Import matplotlib for graph visualization
import matplotlib.pyplot as plt

# Import seaborn for statistical graphs
import seaborn as sns

# Import train_test_split for splitting the dataset
from sklearn.model_selection import train_test_split

# Import Random Forest Machine Learning algorithm
from sklearn.ensemble import RandomForestClassifier

# Import evaluation metrics
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

# Import os for file and folder operations
import os

# Create a folder to store project results
os.makedirs("results", exist_ok=True)

print("All libraries imported successfully!")
```

```python
# Load the uploaded dataset
df = pd.read_csv("data (1).csv", sep=";")

# Display successful loading message
print("Dataset loaded successfully!")

# Display number of rows and columns
print("Dataset Shape:", df.shape)

# Display the first five records
df.head()
```

```python
# Display dataset information
print("Dataset Information:")

df.info()

# Display number of rows
print("\nNumber of Students:", df.shape[0])

# Display number of columns
print("Number of Columns:", df.shape[1])

# Display column names
print("\nColumn Names:")

print(df.columns.tolist())
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

# Display missing values
print("Missing Values:")

print(missing_values)

# Calculate total missing values
total_missing = df.isnull().sum().sum()

print("\nTotal Missing Values:", total_missing)
```

```python
# Count duplicate records
duplicates = df.duplicated().sum()

# Display duplicate count
print("Duplicate Records:", duplicates)
```

```python
# Count students in each outcome category
outcomes = df["Target"].value_counts()

# Display student outcomes
print("Student Outcomes:")

print(outcomes)

# Display total students
print("\nTotal Students:", df.shape[0])
```

```python
# Create a bar graph for student outcomes
plt.figure(figsize=(8, 5))

df["Target"].value_counts().plot(kind="bar")

# Add graph title
plt.title("Student Outcome Distribution")

# Add axis labels
plt.xlabel("Student Outcome")
plt.ylabel("Number of Students")

# Save graph
plt.savefig("results/student_outcomes.png", bbox_inches="tight")

# Display graph
plt.show()
```

```python
# Create a histogram
plt.figure(figsize=(8, 5))

# Plot student ages
plt.hist(df["Age at enrollment"], bins=15)

# Add graph title
plt.title("Student Age Distribution")

# Add axis labels
plt.xlabel("Age")
plt.ylabel("Number of Students")

# Save graph
plt.savefig("results/student_age_distribution.png", bbox_inches="tight")

# Display graph
plt.show()
```

```python
# Separate input features
X = df.drop("Target", axis=1)

# Separate target variable
y = df["Target"]

# Display input feature shape
print("Input Features:", X.shape)

# Display target shape
print("Target:", y.shape)
```

```python
# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display training records
print("Training Records:", X_train.shape[0])

# Display testing records
print("Testing Records:", X_test.shape[0])
```

```python
# Create Random Forest Classifier
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

# Train the model
model.fit(X_train, y_train)

# Display training status
print("Random Forest Model Trained Successfully!")
```

```python
# Predict student outcomes
y_pred = model.predict(X_test)

# Display prediction status
print("Prediction Completed!")

# Display first 10 predictions
print("\nPredicted Outcomes:")

print(y_pred[:10])
```

```python
# Calculate model accuracy
accuracy = accuracy_score(y_test, y_pred)

# Display accuracy
print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
# Generate classification report
report = classification_report(
    y_test,
    y_pred,
    zero_division=0
)

# Display report
print("Classification Report:")

print(report)
```

```python
# Calculate confusion matrix
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=model.classes_
)

# Create graph
plt.figure(figsize=(8, 6))

# Display confusion matrix
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

# Add graph title
plt.title("Student Outcome Prediction")

# Add axis labels
plt.xlabel("Predicted Outcome")
plt.ylabel("Actual Outcome")

# Save graph
plt.savefig("results/confusion_matrix.png", bbox_inches="tight")

# Display graph
plt.show()
```

```python
# Extract feature importance values
importance = model.feature_importances_

# Create a DataFrame
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importance
})

# Sort features by importance
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

# Display top 10 important features
print("Top 10 Important Features:")

print(feature_importance.head(10))

# Create graph
plt.figure(figsize=(10, 6))

sns.barplot(
    data=feature_importance.head(10),
    x="Importance",
    y="Feature"
)

# Add graph title
plt.title("Top 10 Important Student Features")

# Save graph
plt.savefig("results/feature_importance.png", bbox_inches="tight")

# Display graph
plt.show()
```

```python
# Create a DataFrame containing model accuracy
accuracy_result = pd.DataFrame({
    "Metric": ["Accuracy"],
    "Value": [accuracy * 100]
})

# Save accuracy as CSV
accuracy_result.to_csv(
    "results/model_accuracy.csv",
    index=False
)

# Display accuracy
print("Model Accuracy:", round(accuracy * 100, 2), "%")

# Display confirmation
print("Accuracy Saved Successfully!")
```

```python
# Save classification report as text file
with open("results/classification_report.txt", "w") as file:

    # Write report
    file.write(report)

# Display confirmation
print("Classification Report Saved Successfully!")
```
